```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

PROJECT_ROOT = Path.cwd().parent
METADATA_PATH = PROJECT_ROOT / "data" / "processed" / "ccd_metadata.csv"

df = pd.read_csv(METADATA_PATH)

print(df.shape)
df.head()
```

```python
print("Dataset shape:")
print(df.shape)

print("\nColumns:")
print(df.columns.tolist())
```

```python
df.isnull().sum()
```

```python
df["weather"].value_counts()
```

```python
df["weather"].value_counts().plot(
    kind="bar",
    figsize=(8, 5),
)

plt.title("Accident Videos by Weather")
plt.xlabel("Weather")
plt.ylabel("Number of Videos")
plt.tight_layout()
plt.show()
```

```python
df["timing"].value_counts()
```

```python
df["timing"].value_counts().plot(
    kind="bar",
    figsize=(7, 5),
)

plt.title("Accident Videos: Day vs Night")
plt.xlabel("Time")
plt.ylabel("Number of Videos")
plt.tight_layout()
plt.show()
```

```python
df["first_accident_frame"].describe()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(
    df["first_accident_frame"].dropna(),
    bins=20,
)

plt.title("Distribution of First Accident Frame")
plt.xlabel("First Accident Frame")
plt.ylabel("Number of Videos")

plt.tight_layout()
plt.show()
```
